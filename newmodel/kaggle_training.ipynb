## 1. Dataset Preparation
**Note:** Please upload your `master_dataset.zip` to Kaggle before running this cell.

```python
# Unzip your dataset
zip_path = '/kaggle/input/your-dataset-name/master_dataset.zip' # Update this!
extract_path = '/kaggle/working/datasets'

if os.path.exists(zip_path):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)
    print("✅ Dataset extracted successfully!")
```

## 2. Configuration
We create a dynamic `data.yaml` that works in the Kaggle environment.

```python
yaml_content = f"""
path: {extract_path}/master_dataset
train: train/images
val: valid/images
test: test/images

nc: 2
names: ['flood', 'water_body']
"""

with open('/kaggle/working/data.yaml', 'w') as f:
    f.write(yaml_content)
print("✅ Configuration YAML created for Kaggle.")
```

## 3. High-Speed Master Training
Launching the Pixel-Perfect Segmentation engine on GPU.

```python
# Initialize Model
try:
    model = YOLO("yolo12s-seg.pt")
    print("✅ Using YOLO12-Seg")
except:
    model = YOLO("yolo11s-seg.pt")
    print("✅ Using YOLO11-Seg Fallback")

# Start Training
model.train(
    data='/kaggle/working/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    device=0, # Use NVIDIA T4
    save=True,
    project='/kaggle/working/training_runs',
    name='rainwise_seg_final',
    overlap_mask=True,
    mask_ratio=1,
    lr0=0.01,
    warmup_epochs=3
)
```
